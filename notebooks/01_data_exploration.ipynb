```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Locate the project root
BASE_DIR = Path.cwd().parent

# Path to the cleaned dataset
DATA_PATH = BASE_DIR / "data" / "clean_dataset.csv"

print("Project directory:", BASE_DIR)
print("Dataset exists:", DATA_PATH.exists())
print("Dataset path:", DATA_PATH)
```

```python
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five rows:")
display(df.head())

print("\nFraud distribution:")
print(df["isFraud"].value_counts())
```

```python
# Count legitimate and fraudulent transactions
fraud_counts = df["isFraud"].value_counts().sort_index()

# Create the bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=["Legitimate", "Fraudulent"],
    y=fraud_counts.values
)

plt.title("Fraud vs Legitimate Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="TransactionAmt",
    bins=50
)

plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="isFraud",
    y="TransactionAmt"
)

plt.title("Transaction Amounts: Fraud vs. Legitimate")
plt.xlabel("Transaction Type (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Transaction Amount")

plt.tight_layout()
plt.show()
```

```python
# Calculate fraud rates by product category
fraud_rates = (
    df.groupby("ProductCD")["isFraud"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

# Create the bar chart
plt.figure(figsize=(9, 5))

sns.barplot(
    x=fraud_rates.index,
    y=fraud_rates.values
)

plt.title("Fraud Rate by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Fraud Rate (%)")

plt.tight_layout()
plt.show()

print("Fraud rates by product category:")
print(fraud_rates.round(2))
```

```python
# Select a reproducible sample for visualization
sample_df = df.sample(n=5000, random_state=42)

# Create scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=sample_df,
    x="TransactionDT",
    y="TransactionAmt",
    hue="isFraud",
    alpha=0.6
)

plt.title("Transaction Amount vs Transaction Time")
plt.xlabel("Transaction Time (elapsed units)")
plt.ylabel("Transaction Amount")

plt.tight_layout()
plt.show()
```

```python
# Select numerical features for correlation analysis
selected_features = [
    "TransactionAmt",
    "TransactionDT",
    "card1",
    "card2",
    "addr1",
    "dist1",
    "isFraud"
]

# Calculate correlations
correlation_matrix = df[selected_features].corr()

# Plot the heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Selected Features")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Locate the project directory
BASE_DIR = Path.cwd().parent

# Path to the cleaned dataset
DATA_PATH = BASE_DIR / "data" / "clean_dataset.csv"

# Load the dataset
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nFirst five rows:")
display(df.head())

print("\nFraud distribution:")
print(df["isFraud"].value_counts())
```

```python
# DAY 5 - STEP 2: Statistical Analysis

# Separate legitimate and fraudulent transactions
legitimate = df[df["isFraud"] == 0]["TransactionAmt"]
fraudulent = df[df["isFraud"] == 1]["TransactionAmt"]

# Calculate statistics
statistics = pd.DataFrame({
    "Metric": [
        "Mean",
        "Median",
        "Variance",
        "Standard Deviation"
    ],
    "Legitimate Transactions": [
        legitimate.mean(),
        legitimate.median(),
        legitimate.var(),
        legitimate.std()
    ],
    "Fraudulent Transactions": [
        fraudulent.mean(),
        fraudulent.median(),
        fraudulent.var(),
        fraudulent.std()
    ]
})

print("Transaction Amount Statistics")
display(statistics.round(2))
```

```python
# DAY 5 - STEP 3: Transaction Amount Distributions

import matplotlib.pyplot as plt
import seaborn as sns

# Create two plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Legitimate transaction amounts
sns.histplot(
    data=df[df["isFraud"] == 0],
    x="TransactionAmt",
    bins=50,
    ax=axes[0]
)

axes[0].set_title("Legitimate Transaction Amounts")
axes[0].set_xlabel("Transaction Amount")
axes[0].set_ylabel("Number of Transactions")

# Fraudulent transaction amounts
sns.histplot(
    data=df[df["isFraud"] == 1],
    x="TransactionAmt",
    bins=50,
    ax=axes[1]
)

axes[1].set_title("Fraudulent Transaction Amounts")
axes[1].set_xlabel("Transaction Amount")
axes[1].set_ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

```python
# DAY 5 - STEP 4: Explore Important Features

# Select useful columns for our initial analysis
features = [
    "TransactionAmt",
    "TransactionDT",
    "ProductCD",
    "card1",
    "card2",
    "isFraud"
]

# Display the selected columns
display(df[features].head())

# Check the data types
print("\nFeature data types:")
print(df[features].dtypes)

# Check missing values
print("\nMissing values:")
print(df[features].isnull().sum())
```

```python
import numpy as np

# Create a copy so we don't modify the original dataframe
feature_df = df.copy()

# Feature 1: Reduce the skew of transaction amounts
feature_df["TransactionAmt_log"] = np.log1p(
    feature_df["TransactionAmt"]
)

# Feature 2: Extract a time-related feature
feature_df["TransactionHour"] = (
    (feature_df["TransactionDT"] // 3600) % 24
)

# Display the original and new features
display(
    feature_df[
        [
            "TransactionAmt",
            "TransactionAmt_log",
            "TransactionDT",
            "TransactionHour",
            "isFraud"
        ]
    ].head(10)
)

print("Original dataset shape:", df.shape)
print("Feature dataset shape:", feature_df.shape)
```

```python
# Compare skewness before and after log transformation

original_skewness = feature_df["TransactionAmt"].skew()

log_skewness = feature_df["TransactionAmt_log"].skew()

print("Original transaction amount skewness:",
      round(original_skewness, 2))

print("Log-transformed amount skewness:",
      round(log_skewness, 2))

if abs(log_skewness) < abs(original_skewness):
    print("\nThe log transformation reduced the skewness.")
else:
    print("\nThe log transformation did not reduce the absolute skewness.")
```

```python
# Calculate fraud rate for each derived transaction hour

hourly_fraud_rate = (
    feature_df.groupby("TransactionHour")["isFraud"]
    .mean()
    .mul(100)
)

print("Fraud rate by derived transaction hour:")
print(hourly_fraud_rate.round(2))

# Plot the results
plt.figure(figsize=(12, 5))

sns.lineplot(
    x=hourly_fraud_rate.index,
    y=hourly_fraud_rate.values,
    marker="o"
)

plt.title("Fraud Rate by Derived Transaction Hour")
plt.xlabel("Derived Transaction Hour (0–23)")
plt.ylabel("Fraud Rate (%)")
plt.xticks(range(24))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Count transactions and fraudulent transactions per derived hour

hourly_analysis = feature_df.groupby("TransactionHour").agg(
    total_transactions=("isFraud", "count"),
    fraud_transactions=("isFraud", "sum"),
    fraud_rate=("isFraud", "mean")
)

hourly_analysis["fraud_rate"] *= 100

print("Transaction counts and fraud rates by derived hour:")

display(hourly_analysis.round(2))
```

```python
# Validate the engineered features

print("Dataset shape:", feature_df.shape)

print("\nMissing values:")
print(
    feature_df[
        ["TransactionAmt_log", "TransactionHour"]
    ].isnull().sum()
)

print("\nTransaction amount log statistics:")
print(feature_df["TransactionAmt_log"].describe().round(2))

print("\nDerived transaction hour values:")
print(sorted(feature_df["TransactionHour"].unique()))

print("\nInvalid transaction hours:")
print(
    (~feature_df["TransactionHour"].between(0, 23)).sum()
)
```

```python
# DAY 5 - FINAL VALIDATION

print("Dataset shape:", feature_df.shape)

print("\nMissing values:")
print(
    feature_df[
        ["TransactionAmt_log", "TransactionHour"]
    ].isnull().sum()
)

print("\nUnique transaction hours:")
print(sorted(feature_df["TransactionHour"].unique()))

print("\nInvalid transaction hours:")
print(
    (~feature_df["TransactionHour"].between(0, 23)).sum()
)
```
